```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
missing_plot = (
    df.isnull()
    .sum()
    .to_frame("Missing_Count")
)

missing_plot["Missing_Percentage"] = (
    missing_plot["Missing_Count"] / len(df) * 100
)

missing_plot = missing_plot.sort_values(
    "Missing_Percentage",
    ascending=False
)

plt.figure(figsize=(12, 6))

plt.bar(
    missing_plot.index,
    missing_plot["Missing_Percentage"]
)

plt.xticks(rotation=45)
plt.ylabel("Missing Percentage (%)")
plt.title("Missing Values by Column")
plt.tight_layout()
plt.show()
```

```python
df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"],
    format="mixed",
    errors="coerce"
)
```

```python
df["InvoiceDate"].isna().sum()
```

```python
df["Date"] = df["InvoiceDate"].dt.date

df["Year"] = df["InvoiceDate"].dt.year

df["MonthNumber"] = df["InvoiceDate"].dt.month

df["MonthName"] = df["InvoiceDate"].dt.month_name()

df["Quarter"] = df["InvoiceDate"].dt.quarter

df["Day"] = df["InvoiceDate"].dt.day

df["DayName"] = df["InvoiceDate"].dt.day_name()

df["Hour"] = df["InvoiceDate"].dt.hour
```

```python
df["MonthYear"] = df["InvoiceDate"].dt.to_period("M").astype(str)
```

```python
df.duplicated().sum()
```

```python
duplicates = df[df.duplicated(keep=False)]

duplicates.head()
```

```python
df = df.drop_duplicates()
```

```python
df["transactionaltype"].value_counts().plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Sales vs Cancellations")
plt.ylabel("Number of Transactions")
plt.show()
```

```python
(df["Quantity"] < 0).sum()
```

```python
df[df["transactionaltype"] == "Cancellation"][
    ["InvoiceNo", "Quantity", "transactionaltype"]
].head()
```

```python
df["SalesQty"] = np.where(
    df["transactionaltype"] == "Sale",
    df["Quantity"],
    0
)
```

```python
df["CancellationQty"] = np.where(
    df["transactionaltype"] == "Cancellation",
    abs(df["Quantity"]),
    0
)
```

```python
df["UnitPrice"].nunique()
```

```python
df["UnitPrice"].unique()
```

```python
df["Revenue"] = df["Quantity"] * df["UnitPrice"]
```

```python
df["SalesRevenue"] = np.where(
    df["transactionaltype"] == "Sale",
    df["Quantity"] * df["UnitPrice"],
    0
)
```

```python
df["CancellationValue"] = np.where(
    df["transactionaltype"] == "Cancellation",
    abs(df["Quantity"] * df["UnitPrice"]),
    0
)
```

```python
df["CustomerID"].isna().sum()
```

```python
customer_df = df[
    df["CustomerID"].notna()
].copy()
```

```python
df["CustomerStatus"] = np.where(
    df["CustomerID"].isna(),
    "Unknown Customer",
    "Identified Customer"
)
```

```python
total_orders = df["InvoiceNo"].nunique()

print("Total Orders:", total_orders)
```

```python
total_quantity = df["SalesQty"].sum()

print("Total Products Sold:", total_quantity)
```

```python
total_cancellations = (
    df[df["transactionaltype"] == "Cancellation"]
    ["InvoiceNo"]
    .nunique()
)

print(total_cancellations)
```

```python
cancellation_rate = (
    total_cancellations /
    total_orders
) * 100

print(cancellation_rate)
```

```python
country_analysis = (
    df.groupby("Country")
    .agg(
        Orders=("InvoiceNo", "nunique"),
        Quantity=("SalesQty", "sum"),
        Customers=("CustomerID", "nunique")
    )
    .sort_values("Orders", ascending=False)
)

country_analysis.head(10)
```

```python
country_analysis.head(10)["Orders"].plot(
    kind="bar",
    figsize=(10,6)
)

plt.title("Top 10 Countries by Orders")
plt.ylabel("Orders")
plt.show()
```

```python
product_analysis = (
    df.groupby(
        ["StockCode", "Description"]
    )
    .agg(
        Quantity=("SalesQty", "sum"),
        Orders=("InvoiceNo", "nunique")
    )
    .sort_values(
        "Quantity",
        ascending=False
    )
)
```

```python
product_analysis.head(10)
```

```python
customer_analysis = (
    customer_df.groupby("CustomerID")
    .agg(
        Orders=("InvoiceNo", "nunique"),
        Quantity=("SalesQty", "sum"),
        Countries=("Country", "nunique")
    )
    .sort_values(
        "Orders",
        ascending=False
    )
)
```

```python
monthly_sales = (
    df.groupby("MonthYear")
    .agg(
        Orders=("InvoiceNo", "nunique"),
        Quantity=("SalesQty", "sum")
    )
)
```

```python
monthly_sales["Quantity"].plot(
    figsize=(12,6)
)

plt.title("Monthly Sales Quantity")
plt.xlabel("Month")
plt.ylabel("Quantity")
plt.xticks(rotation=45)
plt.show()
```

```python
day_analysis = (
    df.groupby("DayName")
    .agg(
        Orders=("InvoiceNo", "nunique"),
        Quantity=("SalesQty", "sum")
    )
)
```

```python
day_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

day_analysis = day_analysis.reindex(day_order)
```

```python
hour_analysis = (
    df.groupby("Hour")
    .agg(
        Orders=("InvoiceNo", "nunique"),
        Quantity=("SalesQty", "sum")
    )
)
```

```python
hour_analysis["Orders"].plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Orders by Hour")
plt.xlabel("Hour")
plt.ylabel("Orders")
plt.show()
```

```python
numeric_cols = [
    "Quantity",
    "SalesQty",
    "CancellationQty",
    "CustomerID"
]

corr = df[numeric_cols].corr()

plt.figure(figsize=(8,6))
sns.heatmap(
    corr,
    annot=True
)

plt.title("Correlation Matrix")
plt.show()
```
